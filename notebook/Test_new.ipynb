```python
import os
import random

def get_first_token(input_string):
    tokens = input_string.split("_")
    if tokens:
        return tokens[0]
    return None
    
def get_second_token(input_string):
    tokens = input_string.split("_")
    if tokens:
        return tokens[1]
    return None
    
# Set the directory path
directory_path = "AudioArchive/TestAudioShort/"
splitted_dir = 'AudioArchive/TestSplitted'
waveletname = "mexh" 
correct = 0
errors = 0

# Get a list of all files in the directory
file_list = [f for f in os.listdir(directory_path) if os.path.isfile(os.path.join(directory_path, f))]
# Check if there are any files in the directory
if not file_list:
    print("No files found in the directory.")
else:
    # Iterate directory
    for filename in file_list:
        # check if current path is a file
        filenamep = directory_path+filename
        print("Selected file:", filenamep)
        mykey = os.path.splitext(filename)[0]
        #print(mykey)
        # Get and print the first token
        first_token = get_first_token(mykey)
        #print("first_token", first_token)
        # Detection, split e prediction
        peak_detection_and_split(filenamep, waveletname, splitted_dir)
        st = predictOne(splitted_dir, first_token)
        if first_token is not None:
            #print("Real value:", first_token.lower())
            #print("Prediction:", st.lower())
            if(first_token.lower() == st.lower()):
                correct = correct + 1
                print("Correct!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")
            else:
                errors = errors + 1
                print("Error!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!")
```
