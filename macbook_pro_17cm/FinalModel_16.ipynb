```python
# Load the audio file
input_wav = "archive/fold16/output/"
output_wav = "archive/fold16/spectrograms/"
# Iterate through input files
for file_name in os.listdir(input_wav):
    if file_name.endswith(".wav"):
        file_path = os.path.join(input_wav, file_name)
        y, sr = librs.load(file_path)
        # Compute the spectrogram
        D = librs.amplitude_to_db(np.abs(librs.stft(y)), ref=np.max)
        # Plot the spectrogram
        plt.figure(figsize=(12, 8))
        plt.axis('off')
        librs.display.specshow(D, sr=sr)
        spect_name = os.path.splitext(file_name)[0]
        print(spect_name)
        plt.savefig(output_wav+"/"+spect_name+".png")
        plt.close()
print("Spectrogram creation completed!")
```
