```python
import os
import csv
```

```python
def list_files_in_directory(directory):
    file_list = []
    for filename in os.listdir(directory):
        if os.path.isfile(os.path.join(directory, filename)):
            file_list.append(filename)
    return file_list
```

```python
def is_valid_char(char):
    if len(char) == 1 and char.isalpha():
        return True
    else:
        return False
#input_char = input("Enter a single character: ")
#if is_valid_char(input_char):
#    print("True")
#else:
#    print("False")
```

```python
def char_to_value(char):
    if 'A' <= char <= 'Z':
        return ord(char) - ord('A') + 11
    else:
        return char

input_string = "Z"
char_to_value(input_string)
#result = ''.join(str(char_to_value(c)) for c in input_string)
#print(result)
```

```python
def write_to_csv(file_list, output_filename):
    with open(output_filename, 'w', newline='') as csvfile:
        csv_writer = csv.writer(csvfile)
        csv_writer.writerow(['File Name'])
        for filename in file_list:
            input_string = filename
            substrings = input_string.split('_')
            c = 0
            x = 0
            y = 0
            for substring in substrings:
                #print(substring)
                if (c == 1):
                    if is_valid_char(substring):
                        y = (substring)
                        #print(y)
                        x = int(char_to_value(substring))
                        #print(x)
                    else:
                        y = int(substring)
                        x = int(substring) + 1
                c += 1
            mystr = filename+",16,"+str(x)+","+str(y)
            csv_writer.writerow([mystr])
```

```python
input_directory = "archive/fold16/output/"
output_filename = "archive/list.csv"
files_list = list_files_in_directory(input_directory)
write_to_csv(files_list, output_filename)
print(f"File list has been written to {output_filename}")
```
