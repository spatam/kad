```python
import soundfile as sf
import os
from pydub import AudioSegment
from pydub.silence import split_on_silence
import matplotlib.pyplot as plt
import librosa
import numpy as np
```

```python
def split_audio(file_name_without_extension, input_file, output_dir, test_dir, num_splits):
    audio = AudioSegment.from_wav(input_file)
    duration = len(audio)
    split_duration = duration // num_splits

    for i in range(num_splits):
        start_time = i * split_duration
        end_time = (i + 1) * split_duration if i != num_splits - 1 else duration
        split_audio = audio[start_time:end_time]
        output_filename = os.path.join(output_dir, f"{file_name_without_extension}_split_{i + 1}.wav")
        split_audio.export(output_filename, format="wav")
        print(f"Split {i + 1} saved as {output_filename}")
        if(i == 5):
            output_filename = os.path.join(test_dir, f"{file_name_without_extension}_split_{i + 1}.wav")
            split_audio.export(output_filename, format="wav")
            print(f"Split {i + 1} saved as {output_filename}")
```

```python
input_directory = "archive/fold16/input"  # Replace with the path to your input directory
output_directory = "archive/fold16/output_directory"  # Replace with the path to your output directory
test_directory = "archive/test/normal"  # Replace with the path to your output directory
silence_threshold = -10  # Adjust this threshold as needed
num_splits = 25
   
if not os.path.exists(output_directory):
    os.makedirs(output_directory)

for filename in os.listdir(input_directory):
    if filename.endswith(".wav"):
        input_file = os.path.join(input_directory, filename)
        file_name_without_extension = os.path.splitext(os.path.basename(input_file))[0]
        split_audio(file_name_without_extension, input_file, output_directory, test_directory, num_splits)
```

```python
input_wav = "./archive/fold16/output_directory/a_split_10.wav"
y, sr = librosa.load(input_wav)
onset_env = librosa.onset.onset_strength(y=y, sr=sr, hop_length=512, aggregate=np.median)
peaks = librosa.util.peak_pick(onset_env, pre_max=3, post_max=3, pre_avg=3, post_avg=5, delta=0.5, wait=10)
peaks
```

```python
plt.figure(figsize=(10, 6))
times = librosa.times_like(onset_env, sr=sr, hop_length=512)
fig, ax = plt.subplots(nrows=2, sharex=True)
D = np.abs(librosa.stft(y))
librosa.display.specshow(librosa.amplitude_to_db(D, ref=np.max), y_axis='log', x_axis='time', ax=ax[1])
ax[0].plot(times, onset_env, alpha=0.8, label='Onset strength')
ax[0].vlines(times[peaks], 0, onset_env.max(), color='r', alpha=0.8, label='Selected peaks')
ax[0].legend(frameon=True, framealpha=0.8)
ax[0].label_outer()
plt.savefig("results/a_1_peak_01_secs.png")
```
